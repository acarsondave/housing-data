```
def trellis(web_url):
  url = "https://trellis.law/cases/%s?state=ca&state=de&state=fl&state=il&state=ny&state=tx" %(web_url)
  driver.get(url)
  def login():
    driver.get("https://trellis.law/login")
    email = driver.find_element_by_id("email")
    email.send_keys("yatang.lin@gmail.com")
    password = driver.find_element_by_id("password")
    password.send_keys("tuzihulilove")
    submit = driver.find_element_by_class_name("submitButton___hEn03")
    submit.click()
    time.sleep(2)
    print("Logged in...")
    driver.get(url)
    time.sleep(1)
  try:
    check = len(driver.find_elements_by_class_name("btn-account-text"))
    print(check)
    if check == 0:
        print("Not logged in")
        login()
    else:
        print("Logged in...")
  except Exception as e:
    print(e)
  l = []
  def getinfo():
    soup = BeautifulSoup(driver.page_source,"html.parser")
    card = soup.find_all("div",{"class":"cardInfo"})
    for items in range(len(card)):
      left = card[items].find("div",{"class":"cardLeftInfo"}).text
      right = card[items].find("div",{"class":"cardRightInfo"}).text
      lines = left.split("\n")
      non_empty_lines = [line for line in lines if line.strip() != ""]
      string_without_empty_lines = ""

      for line in non_empty_lines:
            string_without_empty_lines += line + "\n"
      leftinfo = string_without_empty_lines.split("\n")
      leftinfo = list(filter(None, leftinfo))

      lines = right.split("\n")
      non_empty_lines = [line for line in lines if line.strip() != ""]
      string_without_empty_lines = ""

      for line in non_empty_lines:
            string_without_empty_lines += line + "\n"
      rightinfo = string_without_empty_lines.split("\n")
      rightinfo = list(filter(None, rightinfo))
      d = {}
      d["Name"] = driver.find_elements_by_class_name("justify-content-start")[items+1].text
      x = 0
      for i in range(int(len(leftinfo)/2)):
        d[leftinfo[x]] = leftinfo[x+1]
        x +=2

      x = 0
      for i in range(int(len(rightinfo)/2)):
        d[rightinfo[x]] = rightinfo[x+1]
        x +=2

      l.append(d)
  while True:
    try:
        try:
          getinfo()
          element = driver.find_element_by_link_text("next")
          print(element.text)
          driver.execute_script("arguments[0].scrollIntoView();", element)
          element.click()
          print(driver.current_url)
        except NoSuchElementException as e:
          print("finished...")
          sys.exit()
    except SystemExit:
      break
  df = pandas.DataFrame(l)
  csv = "trellis-%s.csv" %(web_url)
  df.to_csv(csv,index=False)
  print("finished " + csv)

trellis("uber")
```
